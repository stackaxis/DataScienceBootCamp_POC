```python
### Assignment 2: Function with Nested Default Arguments

"""Define a function that takes two arguments, a and b, where b is a dictionary with a default value of an empty dictionary. 
The function should add a new key-value pair to the dictionary and return it. Test the function with different inputs."""

def nest_function(a,b={}):
    b[a]=a
    return b

# Test the function with different inputs
print(nest_function("key1"))
print(nest_function("key2"))
print(nest_function("key3"))

# Above code will return the same dictionary for each call, which is not the desired behavior.
# above code shows that the default argument b is mutable and retains its state across function calls.Below code shows the correct way to handle mutable default arguments by creating a new dictionary for each call.
#it solve  mutable default argument issue by using None as the default value for b and creating a new dictionary inside the function if b is None.

##  Below creates new dictionary for each call, which is the correct way to handle mutable default arguments.

def nest_function(a, b=None):
    if b is None:
        b = {}
    b[a]=a
    return b

# Test
print(nest_function("key1"))
print(nest_function("key2"))
print(nest_function("key3"))
```

```python
# Assignment1: fibonacci Series with Memoization

def fibonacci(n,memo={}):
    if n <=1:
        return n
    if n in memo:
        return memo[n]
    
    memo[n]=fibonacci(n-1,memo) + fibonacci(n-2,memo)
    return memo[n]

# Test
print(fibonacci(10))  # 55
print(fibonacci(15))  # 610
```

```python
### Assignment 3: Function with Variable Keyword Arguments

"""Define a function that takes a variable number of keyword arguments and returns a dictionary containing only those key-value pairs 
where the value is an integer. Test the function with different inputs."""

def filter_integers(**kwargs):
    result = {}
    for k,v in kwargs.items():
        if isinstance(v,int):
            result[k] = v
    return result

# Test the function with different inputs
print(filter_integers(a=1, b="string", c=3.14, d=5))  # {'a': 1}
print(filter_integers(x=10, y=20, z="hello"))  # {'x': 10}
print(filter_integers(name="Alice", age=30, city="New York"))  # {'age': 30}


# dictionary comprehension is a concise way to create dictionaries in Python. It allows you to generate a new dictionary by applying 
# an expression to each item in an iterable, optionally filtering items based on a condition. The syntax for dictionary comprehension is 
# similar to list comprehension, but it uses curly braces {} instead of square brackets [].

def filter_integers(**kwargs):
    return {k: v for k, v in kwargs.items() if isinstance(v, int)}

# Test
print(filter_integers(a=1, b='two', c=3, d=4.5))  # {'a': 1, 'c': 3}
print(filter_integers(x=10, y='yes', z=20))  # {'x': 10, 'z': 20}
```

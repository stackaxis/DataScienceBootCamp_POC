```python
### Assignment 2: Function with Nested Default Arguments

"""Define a function that takes two arguments, a and b, where b is a dictionary with a default value of an empty dictionary. 
The function should add a new key-value pair to the dictionary and return it. Test the function with different inputs."""

def nest_function(a,b={}):
    b[a]=a
    return b

# Test the function with different inputs
print(nest_function("key1"))
print(nest_function("key2"))
print(nest_function("key3"))

# Above code will return the same dictionary for each call, which is not the desired behavior.
# above code shows that the default argument b is mutable and retains its state across function calls.Below code shows the correct way to handle mutable default arguments by creating a new dictionary for each call.
#it solve  mutable default argument issue by using None as the default value for b and creating a new dictionary inside the function if b is None.

##  Below creates new dictionary for each call, which is the correct way to handle mutable default arguments.

def nest_function(a, b=None):
    if b is None:
        b = {}
    b[a]=a
    return b

# Test
print(nest_function("key1"))
print(nest_function("key2"))
print(nest_function("key3"))
```

```python
# Assignment1: fibonacci Series with Memoization

def fibonacci(n,memo={}):
    if n <=1:
        return n
    if n in memo:
        return memo[n]
    
    memo[n]=fibonacci(n-1,memo) + fibonacci(n-2,memo)
    return memo[n]

# Test
print(fibonacci(10))  # 55
print(fibonacci(15))  # 610
```

```python
### Assignment 3: Function with Variable Keyword Arguments

"""Define a function that takes a variable number of keyword arguments and returns a dictionary containing only those key-value pairs 
where the value is an integer. Test the function with different inputs."""

def filter_integers(**kwargs):
    result = {}
    for k,v in kwargs.items():
        if isinstance(v,int):
            result[k] = v
    return result

# Test the function with different inputs
print(filter_integers(a=1, b="string", c=3.14, d=5))  # {'a': 1}
print(filter_integers(x=10, y=20, z="hello"))  # {'x': 10}
print(filter_integers(name="Alice", age=30, city="New York"))  # {'age': 30}


# dictionary comprehension is a concise way to create dictionaries in Python. It allows you to generate a new dictionary by applying 
# an expression to each item in an iterable, optionally filtering items based on a condition. The syntax for dictionary comprehension is 
# similar to list comprehension, but it uses curly braces {} instead of square brackets [].

def filter_integers(**kwargs):
    return {k: v for k, v in kwargs.items() if isinstance(v, int)}

# Test
print(filter_integers(a=1, b='two', c=3, d=4.5))  # {'a': 1, 'c': 3}
print(filter_integers(x=10, y='yes', z=20))  # {'x': 10, 'z': 20}
```

```python
### Assignment 4: Function with Callback

"""Define a function that takes another function as a callback and a list of integers. 
The function should apply the callback to each integer in the list and return a new list with the results. 
Test with different callback functions."""

def apply_callback(callback,int_list):
    return [callback(x) for x in int_list]

# Test the function with different callback functions
print(apply_callback(lambda x:x**2,[1,2,3,4]))
print(apply_callback(lambda x:x*2,[5,6,7,8]))
```

```python
### Assignment 6: Function with Decorators

''''Define a function that calculates the time taken to execute another function. Apply this decorator to a function that performs a complex calculation. Test the decorated function with different inputs.'''
import time
def time_decorator(func):
    def wrapper(*args, **kwargs):
        start_time = time.time()
        result=func(*args, **kwargs)
        end_time = time.time()
        print(f"Time Taken to execute {func.__name__}: {end_time - start_time} seconds")
        return result
    return wrapper

@time_decorator
def complex_calculation(n):
    total = 0
    for i in range(n):
        total += i**2
    return total

print(complex_calculation(1000000))
```

```python
### Assignment 7: Higher-Order Function for Filtering and Mapping

"""Define a higher-order function that takes two functions, a filter function and a map function, along with a list of integers. The higher-order function should first filter the integers using the filter function and then apply the map function to the filtered integers. Test with different filter and map functions."""


def higher_order_function(filter_func, map_func, int_list):
    filtered_list=list(filter(filter_func, int_list))
    mapped_list=list(map(map_func, filtered_list))
    return mapped_list

# Test the higher-order function with different filter and map functions
print(higher_order_function(lambda x: x % 2 == 0, lambda x: x**2, [1, 2, 3, 4, 5]))  # [4, 16]
print(higher_order_function(lambda x: x > 3, lambda x: x *  2, [1, 2, 3, 4, 5]))  # [8, 10]
```

```python
### Assignment 8: Function Composition

"""Define a function that composes two functions, f and g, such that the result is f(g(x)). Test with different functions f and g."""

def composite_function(f,g):
    def composed(x):
        return f(g(x))
    return composed


# Test the composite function with different functions f and g
def square(x):
    return x ** 2

def double(x):
    return 2 * x

# Compose the functions
composed_func = composite_function(square, double)

# Test the composed function
print(composed_func(3))  # Should print 36 (square of double of 3)


#### 2. Another Sample of function composition using lambda functions

def composite_function(f, g):
    return lambda x:f(g(x))

# Test the composite function with different functions f and g
f= lambda x:x**2
g= lambda x:2*x
h = composite_function(f, g)
print(h(3))  # Should print 36 (square of double of 3)
print(h(5))  # Should print 100 (square of double of 5)
```
